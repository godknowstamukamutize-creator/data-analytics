# Exercise 2.C Creating and calling functions
## Lab 1. Working with defining custom functions

```python
def favorite_things(name, movie, music):
    return f"Name: {name}\nFavorite Movie: {movie}\nFavorite Musician/Band: {music}"

print(favorite_things('Godknows Mutize', 'Vikings', 'Maverick City'))
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here('I joined Year Up United to build my data analytics skills and launch a tech career and be able get opportunities.'))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place('Church', 'its a place I got to worship God'))
```

```python
print(favorite_place('the beach', 'it helps me relax and clear my mind'))
```
