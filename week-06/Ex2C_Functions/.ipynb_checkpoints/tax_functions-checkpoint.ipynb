# Exercise 2.C Lab 3 – Tax Functions
This notebook demonstrates custom functions that calculate payroll tax deductions,
including Social Security, Medicare, and Federal withholding based on a withholding code.

```python
def get_soc_sec_tax(gross_pay):
    rate = 0.062  # 6.2%
    return gross_pay * rate
```

```python
def get_medicare_tax(gross_pay):
    rate = 0.0145  # 1.45%
    return gross_pay * rate
```

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:
        rate = 0.185
    else:
        # Code 4+: 18% minus 0.5% for each withholding over 4
        extra = withholding_code - 4
        rate = 0.18 - (extra * 0.005)
    return gross_pay * rate
```

```python
p1_gross = 750
p1_code = 0

p1_ss = get_soc_sec_tax(p1_gross)
p1_med = get_medicare_tax(p1_gross)
p1_fed = get_federal_tax(p1_gross, p1_code)

print(f"Person 1 | Gross: ${p1_gross:.2f} | Withholding Code: {p1_code}")
print(f"  Social Security Tax: ${p1_ss:.2f}")
print(f"  Medicare Tax:        ${p1_med:.2f}")
print(f"  Federal Tax:         ${p1_fed:.2f}")
```

```python
p2_gross = 1550
p2_code = 2

p2_ss = get_soc_sec_tax(p2_gross)
p2_med = get_medicare_tax(p2_gross)
p2_fed = get_federal_tax(p2_gross, p2_code)

print(f"Person 2 | Gross: ${p2_gross:.2f} | Withholding Code: {p2_code}")
print(f"  Social Security Tax: ${p2_ss:.2f}")
print(f"  Medicare Tax:        ${p2_med:.2f}")
print(f"  Federal Tax:         ${p2_fed:.2f}")
```
