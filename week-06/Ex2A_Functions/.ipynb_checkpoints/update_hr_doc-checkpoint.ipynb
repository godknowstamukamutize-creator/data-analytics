## Lab 3. Employee data
This lab exercise makes updates to
 
information extracted from
an employee database.

```python
hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'),
    ('0234', 'OPS', 'Jim Smith', '54000')
]
```

```python
# Tuples are immutable, so replace the whole tuple
hr_list[1] = (hr_list[1][0], hr_list[1][1], 'Mark Blick-Hawley', hr_list[1][3])
```

```python
print(hr_list)
```

```python
hr_list[3] = (hr_list[3][0], 'CS', hr_list[3][2], '60000')
```

```python
print(hr_list)
```

```python
for emp in hr_list:
    salary_formatted = f"{int(emp[3]):,}"
    print(f"{emp[0]} | {emp[1]} | {emp[2]} | {salary_formatted}")
```
