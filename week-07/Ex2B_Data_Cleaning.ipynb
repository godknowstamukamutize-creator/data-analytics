# Exercise 2.B Cleaning up data

```python
import numpy as np
import pandas as pd
```

```python
# Loading test_data.csv saved from Exercise 2.A
df = pd.read_csv('test_data.csv')
df.describe()
```

## Filtering a DataFrame

```python
# Only rows where Age is greater than 18
df[df['Age'] > 18]
```

```python
# Only rows where Score is greater than 50 AND less than 100
df[(df['Score'] > 50) & (df['Score'] < 100)]
```

## Modifying a DataFrame

```python
# Removing rows where Score is negative (invalid data)
df_cleaned = df.drop(df[df['Score'] < 0].index)
df_cleaned
```

```python
# Removing the City column
df_cleaned = df_cleaned.drop(columns=['City'])
df_cleaned
```

```python
# Replacing missing Age values with the mean Age
df_cleaned['Age'] = df_cleaned['Age'].fillna(df_cleaned['Age'].mean())
df_cleaned
```

```python
# Replacing any Score value of 0 with the mean Score
df_cleaned['Score'] = df_cleaned['Score'].replace(0, df_cleaned['Score'].mean())
df_cleaned
```

```python
# Dropping any remaining rows that still contain NaN values
df_cleaned = df_cleaned.dropna()
df_cleaned
```

```python
# Apply ID values as the index, then drop the ID column
# Step 1: setting the index to the ID column
df_cleaned = df_cleaned.set_index('ID')
# Step 2: ID is now the index label, so no separate drop needed
df_cleaned
```
