# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
# Create a 2D array with two rows: (11,12,13) and (14,15,16)
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
# Loop through the first dimension of arrA — each 'x' is a row (a 1D array)
for x in arrA:
    print(x * 2)
```

```python
# Loop through the second dimension of arrA — each 'y' is a single scalar element
for x in arrA:
    for y in x:
        print(y * 2)
```

Difference: The first loop iterates over the two rows of arrA, printing each row (a 1-D array) multiplied by 2. The second loop iterates over individual elements, so each number is printed on its own line.

```python
# Loop through the second dimension again, but this time assign x*2 to loop_result
for x in arrA:
    for y in x:
        loop_result = x * 2  # x*2 multiplies the whole row, not just y
        print(loop_result)
# loop_result is a full row array each time — y is unused here. 
# loop_result stores x*2, which doubles the entire row, not just the individual element y. 
# This means the full doubled row is printed once per element (6 times total), rather than one doubled value per element.
```

## Joining and Stacking Arrays

```python
# Concatenate arrA with arrA on axis=0 (default) — stacks rows on top of each other
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
# Concatenate arrA with arrA on axis=1 — joins columns side by side
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
# Stack joined_array_1 with itself on the default axis (axis=0)
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
# Stack joined_array_1 with joined_array_2 
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2)

# ValueError: all input arrays must have the same shape
# joined_array_1 is (4,3) and joined_array_2 is (2,6) — shapes don't match
```

```python
# Concatenate joined_array_1 with arrA on the default axis (axis=0)
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
# Concatenate joined_array_1 with arrA on axis=1
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
print(another_2)

# When joining on axis=1 (side by side), numpy requires both arrays to have the
# same number of ROWS. joined_array_1 has 4 rows but arrA only has 2 rows,
# so numpy can't line them up column-by-column — the row counts don't match.
```

```python
# hstack: horizontal stack — joins arrays side by side (along axis=1 for 2D arrays)
hstacked = np.hstack((arrA, arrA))
print("hstack result:")
print(hstacked)
# hstack glues arrA to the right of arrA, increasing the number of columns.
# Result shape: (2, 6)

# vstack: vertical stack — joins arrays on top of each other (along axis=0)
vstacked = np.vstack((arrA, arrA))
print("\nvstack result:")
print(vstacked)
# vstack places arrA below arrA, increasing the number of rows.
# Result shape: (4, 3)

# dstack: depth stack — joins arrays along the THIRD axis (depth), creating a 3D array
dstacked = np.dstack((arrA, arrA))
print("\ndstack result:")
print(dstacked)
# dstack layers arrA behind arrA along a new third dimension.
# Result shape: (2, 3, 2)
```

## Splitting Arrays

```python
# Split joined_array_1 into 2 equal parts along axis=0 (default)
splitsky = np.split(joined_array_1, 2)
print(splitsky)

# splitsky is a Python LIST containing 2 arrays.
# Each of those arrays is 2-dimensional with shape (2, 3) — 2 rows and 3 columns.
# The split divided joined_array_1's 4 rows into two groups of 2 rows each.
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
# Result: a Python list of 3 arrays — two with 2 rows, one with 0 rows (empty)
# Each sub-array is still 2-D
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
# Create arrB as a 2D array with rows (7,6), (13,5), (7,3)
arrB = np.array([[7, 6], [13, 5], [7, 3]])

# Sort arrB
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
# Print arrB 
print(arrB)

# Create boolean mask z: True where values in arrB are less than 7
z = arrB < 7

# Apply the mask — arrB[z] returns a 1D array of all values where z is True
arrC = arrB[z]
print(arrC)
```

```python
# Create boolean mask y: True where values in arrB are greater than or equal to 7
y = arrB >= 7

# Apply the mask — arrB[y] returns all values that satisfy the condition
arrD = arrB[y]
print(arrD)
```
