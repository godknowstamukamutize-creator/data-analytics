# Exercise 1.C Performing operations on arrays

```python
import numpy as np
```

## Sample Arrays

```python
# Copied from Exercise 1.B
r_1 = np.random.randint(1, 21, size=2)
r_2 = np.random.randint(1, 21, size=(1, 2))
r_3 = np.random.random(size=(2, 1))
r_4 = np.random.randint(20, 101, size=(4, 5))
r_5 = np.random.random(size=(4, 5))
```

## Permutations

```python
# Print current value of r_4
print(r_4)
```

```python
print(r_1)
```

```python
# Shuffle r_4 in place
np.random.shuffle(r_4)
print(r_4)
```

**Shuffle observation:** `np.random.shuffle` only shuffles along the **x axis** — meaning it rearranges the rows as whole units, but does not change the order of elements within each row. The values inside each row stay together; only which row appears in which position changes.

```python
# Create a permutation of r_4 
new_shuffle = np.random.permutation(r_4)
print('new_shuffle:')
print(new_shuffle)

print('r_4 (unchanged):')
print(r_4)
```

**Shuffle vs Permutation:** Use `shuffle` when we want to randomly reorder an existing array in place and don't need to keep the original order. Use `permutation` when we want a new randomly ordered copy while preserving the original array-useful when we still need to reference the original data after reordering.

## Mathematical Operations

```python
# r_1 + r_2: shape (2,) and (1,2)
print('r_1:', r_1, '| shape:', r_1.shape)
print('r_2:', r_2, '| shape:', r_2.shape)
result = np.add(r_1, r_2)
print('r_1 + r_2:', result)
print('Result shape:', result.shape)
```

```python
# r_1 + r_3: shape (2,) and (2,1)
print('r_1:', r_1, '| shape:', r_1.shape)
print('r_3:', r_3, '| shape:', r_3.shape)
result = np.add(r_1, r_3)
print('r_1 + r_3:', result)
print('Result shape:', result.shape)
```

```python
print('r_2:', r_2, '| shape:', r_2.shape)
print('r_3:', r_3, '| shape:', r_3.shape)
result = np.add(r_2, r_3)
print('r_2 + r_3:', result)
print('result shape:', result.shape)
```

```python
print('r_1:', r_1, '| shape:', r_1.shape)
print('r_4:', r_4, '| shape:', r_4.shape)
result = np.add(r_1, r_4)
print('r_1 + r_4:', result)
print('result shape:', result.shape)

# r_1 cannot broadcast with r_4 because of the different shapes, (2,) and (4,5) respectively.
```

```python
# r_2 + r_4: shape (1,2) and (4,5)
print('r_2:', r_2, '| shape:', r_2.shape)
print('r_4:', r_4, '| shape:', r_4.shape)
result = np.add(r_2, r_4)
print('r_2 + r_4:', result)
print('Result shape:', result.shape)

# r_2 cannot broadcast with r_4 because of the different shapes, (1,2) and (4,5) respectively.
```

```python
print('r_3:', r_3, '| shape:', r_3.shape)
print('r_4:', r_4, '| shape:', r_4.shape)
result = np.add(r_3, r_4)
print('r_3 + r_4:', result)
print('result shape:', result.shape)

# r_3 cannot broadcast with r_4 because of the different shapes, (2,1) and (4,5) respectively.
```

```python
# Add np.array([1]) to r_4 
print(np.add(np.array([1]), r_4))

#single element broadcasts across the entire array
```

```python
# Add np.array([[1]]) to r_4 
print(np.add(np.array([[1]]), r_4))

#2D single element also broadcasts across the entire array
```

```python
# Add np.array([1,1,1,1,1]) to r_4 
print(np.add(np.array([1, 1, 1, 1, 1]), r_4))

#shape (5,) matches 5 columns, broadcasts across all 4 rows
```

```python
# Add np.array([[1],[1]]) to r_4 
print(np.add(np.array([[1],[1]]), r_4))

# ERROR: operands could not be broadcast together with shapes (2,1) and (4,5).
# The first dimension is 2, which does not match r_4's 4 rows and is not 1,
# so numpy cannot stretch it to fit.
```

**How broadcasting works:** When adding two arrays, numpy compares their shapes dimension by dimension from right to left. A dimension is compatible if the sizes are equal, or if one of them is 1, in that case, the size-1 dimension gets "stretched" to match the other. The shape of the result is the larger of the two sizes in each dimension. If any dimension is neither equal nor 1, numpy raises a ValueError. So `np.array([1])` added to a (4,5) array adds 1 to every single element, while `np.array([1,1,1,1,1])` adds element-by-element across each row.

## Rounding

```python
# Print r_3 to see the float values we'll be rounding
print(r_3)
```

```python
# trunc: removes the decimal part, returning the integer portion (toward zero)
print(np.trunc(r_3))
```

```python
# fix: same as trunc — truncates toward zero
print(np.fix(r_3))
```

```python
# floor: rounds DOWN to the nearest integer
print(np.floor(r_3))
```

```python
# ceil: rounds UP to the nearest integer
print(np.ceil(r_3))
```

```python
# round: rounds to 2 decimal places using standard rounding rules
print(np.round(r_3, 2))
```

## Grocery Problem

```python
# Each array represents [vegetables, fruits, snacks, beverages]
bag1 = np.array([2, 3, 4, 0])  # 2 vegetables, 3 fruits, 4 snacks, 0 beverages
bag2 = np.array([4, 0, 1, 2])  # 4 vegetables, 0 fruits, 1 snack, 2 beverages
print('bag1:', bag1)
print('bag2:', bag2)
```

```python
# Total items per bag
total_bag1 = np.sum(bag1)
total_bag2 = np.sum(bag2)
print(f'The first bag contains {total_bag1} items, and the second bag contains {total_bag2} items.')
```

```python
# Total items per category across both bags
category_totals = np.add(bag1, bag2)
print(f'There are {category_totals[0]} vegetables, {category_totals[1]} fruits, '
      f'{category_totals[2]} boxes of snacks, and {category_totals[3]} beverages in the two grocery bags.')
```

## Teamwork

```python
# Ali: Mon=5, Tue=4, Wed=6, Thu=4, Fri=2
ali = np.array([5, 4, 6, 4, 2])
```

```python
# Ben missed Monday — represented as 0; Tue=5, Wed=5, Thu=3, Fri=4
ben = np.array([0, 5, 5, 3, 4])
```

```python
# Carla: Mon=3, Tue=6, Wed=0 (sick), Thu=3, Fri=4
carla = np.array([3, 6, 0, 3, 4])
```

```python
# Stack all three into a single 2D array: each row = one person, each column = one day
all_knocks = np.array([ali, ben, carla])
print(all_knocks)
```

```python
# Total doors knocked per person (sum along axis=1 — across the days)
per_person = np.sum(all_knocks, axis=1)
print(f'Ali knocked on {per_person[0]} doors, Ben on {per_person[1]}, and Carla on {per_person[2]}.')

# Total doors knocked by the whole group
group_total = np.sum(all_knocks)
print(f'The group collectively knocked on {group_total} doors.')
```

```python
# Daily totals for the group 
daily_totals = np.sum(all_knocks, axis=0)

# Cumulative rolling total of doors knocked day by day
rolling_doors = np.cumsum(daily_totals)

# Every 2 door-knocks = 1 estimated vote, so multiply the rolling total by 0.5
estimated_votes = rolling_doors * 0.5

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
for day, votes in zip(days, estimated_votes):
    print(f'{day}: estimated cumulative votes turned out = {int(votes)}')
```
