# Exercise 1.B NumPy Random

```python
import numpy as np
```

## Generating Random Numbers

```python
# Random integers between 1 and 20, 2 elements (1D)
r_1 = np.random.randint(1, 21, size=2)
print(r_1)
```

```python
# Random integers between 1 and 20, 1 row and 2 columns
r_2 = np.random.randint(1, 21, size=(1,2))
print(r_2)
```

```python
# Random floats between 0 and 1, 2 rows and 1 column
r_3 = np.random.random((2, 1))
print(r_3)
```

```python
# Random integers between 20 and 100, 4 rows and 5 columns
r_4 = np.random.randint(20, 101, size=(4, 5))
print(r_4)
```

```python
# Random floats between 0 and 1, 4 rows and 5 columns
r_5 = np.random.random(size=(4, 5))
print(r_5)
```

## Generating Random Choices from an Array

```python
# 1D array of 12 classmate names
students = np.array([
    'Alice', 'Brian', 'Carlos', 'Diana',
    'Evelyn', 'Frank', 'Grace', 'Henry',
    'Isabel', 'James', 'Karen', 'Luis'
])
```

```python
# Choose one random student from the array
call_on = np.random.choice(students)
print(call_on)
```

```python
# Choose one random student from the array
call_on = np.random.choice(students)
print(call_on)
```

```python
# Choose a random sample from students in a 3x4 grid (3 rows, 4 columns = 12 selections)
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)

# This is not an effective way to assign breakout groups because np.random.choice can repeat students,
# so some may appear multiple times while others are left out. 
# A better approach would ensure each student is selected only once using replace=False.
```

```python
# Choose 5 students with weighted probabilities
# Probabilities must sum to 1.0 and there must be one value per student
call_on_weighted = np.random.choice(
    students,
    size=5,
    p=[0.05, 0.30, 0.10, 0.10, 0.10, 0.10, 0.05, 0.10, 0.05, 0.05, 0.00, 0.00]
)
print(call_on_weighted)

# Probability breakdown (p values must sum to 1.0):
# Alice   = 0.05  — low chance
# Brian   = 0.30  — highest chance (30% as required)
# Carlos  = 0.10
# Diana   = 0.10
# Evelyn  = 0.10
# Frank   = 0.10
# Grace   = 0.05  — low chance
# Henry   = 0.10
# Isabel  = 0.05  — low chance
# James   = 0.05  — low chance
# Karen   = 0.00  — zero likelihood of being called on (as required)
# Luis    = 0.00  — zero likelihood of being called on (as required)
```

## Generating a Normal Distribution

```python
# Normal distribution: mean=10, standard deviation=5, size=4x5
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
