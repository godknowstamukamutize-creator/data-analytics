# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, marker='o')
plt.show()
```

```python
plt.plot(x, y, marker='o')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

## Styling Plots

```python
# Dashed line style, red color
plt.plot(x, y, linestyle='--', color='red')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
# Square marker, size 10, green marker color
plt.plot(x, y, linestyle='--', color='red', marker='s', markersize=10, markerfacecolor='green')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
# Triangle-up marker ('^'), purple color
plt.plot(x, y, marker='^', color='purple', markersize=10)
plt.title('Triangle Markers - Purple')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
# Diamond marker ('D'), orange color
plt.plot(x, y, marker='D', color='orange', markersize=10, linestyle='-.')
plt.title('Diamond Markers - Orange')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

```python
# Basic bar chart: product sales
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products, sales)
plt.show()
```

```python
# Customized bar chart: title, axis labels, color, and bar width
plt.bar(products, sales, color='skyblue', width=0.5)
plt.title('Product Sales')
plt.xlabel('Product')
plt.ylabel('Units Sold')
plt.show()
```

```python
# Basic scatter plot with 50 random data points
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)
plt.show()
```

```python
# Customized scatter plot: variable marker size and viridis colormap
s = np.random.rand(50) * 100  # marker size varies per point
c = np.random.rand(50)        # color value per point (mapped through colormap)

plt.scatter(x, y, s=s, c=c, cmap='viridis', alpha=0.7)
plt.colorbar(label='Color value')
plt.title('Customized Scatter Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

## Subplots

```python
# Figure with two subplots side by side: sine wave and histogram
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# --- Subplot 1: Sine wave ---
x_sine = np.linspace(0, 10, 100)
y_sine = np.sin(x_sine)

ax1.plot(x_sine, y_sine, color='steelblue')
ax1.set_title('Sine Wave')
ax1.set_xlabel('X')
ax1.set_ylabel('sin(x)')

# --- Subplot 2: Histogram of 1000 random numbers ---
random_data = np.random.randn(1000)

ax2.hist(random_data, bins=30, color='salmon', edgecolor='white')
ax2.set_title('Histogram of Random Normal Data')
ax2.set_xlabel('Value')
ax2.set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
