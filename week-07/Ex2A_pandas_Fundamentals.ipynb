```python
import numpy as np
import pandas as pd
```

## Series

```python
# Four plain Python objects: list, tuple, set, dictionary
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {'first': 13, 'second': 14, 'third': 15, 'fourth': 16}
```

```python
# Create a pandas Series from each Python object
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(sorted(ac))
ser4 = pd.Series(ad)

print('ser1 (from list):')
print(ser1)
print('\nser2 (from tuple):')
print(ser2)
print('\nser3 (from set):')
print(ser3)
print('\nser4 (from dict):')
print(ser4)
```

```python
# Create a Series from literal integers directly
ser5 = pd.Series(17, 18, 19, 20)
print(ser5)
# TypeError: Index(...) must be called with a collection of some kind, 18 was passed
# pd.Series() expects a single iterable object (like a list), not multiple separate values.
# The correct way would be: pd.Series([17, 18, 19, 20])
```

## DataFrames from Series

```python
# Create a DataFrame from ser1
df1 = pd.DataFrame(ser1)
df1
```

```python
# Add ser2 as column 1 and ser3 as column 2 using insert()
df1.insert(1, 'ser2', ser2)
df1.insert(2, 'ser3', ser3)
df1
```

```python
# Add ser4 as a new column
df1['ser4'] = ser4
df1

# ser4 was created from a dictionary, so its index labels are strings
# ('first', 'second', 'third', 'fourth') instead of integers (0,1,2,3).
# Because those string labels don't match the existing integer row indices in df1,
# pandas cannot align the values — the entire ser4 column shows NaN.
```

```python
# Add a new row at index 4 using .loc
df1.loc[4] = [5, 9, 12, 0]
df1
```

## Reading Data into a DataFrame

```python
# Read sakila_example.csv into df2
df2 = pd.read_csv('sakila_example.csv')
df2
```

```python
# Top 5 records
df2.head()
```

```python
# Last 10 records
df2.tail(10)
```

```python
# Summary information about df2
df2.info()
```

**Facts from info():** The info output tells us how many rows and columns the DataFrame has, the name and data type of each column, and how many non-null values exist per column. Any column with fewer non-null entries than total rows contains missing data. We can also see which columns hold numeric types (int64/float64) versus text (object).

```python
# Column statistics for df2
df2.describe()
```

**Insights from describe():** Useful insights include the min/max range of numeric columns (which can reveal outliers), the mean and standard deviation (which show the typical value and how spread out the data is), and the 25th/75th percentiles (which show where most values fall). Summarizations that are less useful are count and mean for ID-like columns, those are just sequential numbers, so their average or spread has no real meaning.

```python
# Read northwind_example.csv into df3
df3 = pd.read_csv('northwind_example.csv')
df3
```

```python
# Column statistics for df3
df3.describe()
```

**Northwind describe insights:** The northwind data likely contains business metrics like prices, quantities, and order values, so the mean, min/max, and percentiles are genuinely informative, for example, spotting unusually large orders or low-priced products. The difference from the sakila results is that northwind's numeric columns tend to represent real measurable quantities, making statistical summaries more actionable. If northwind has more text/categorical columns, describe() will show fewer columns by default since it only summarizes numeric data.

## Coding a DataFrame

```python
# Create a dictionary with mixed data including NaN and an invalid score (-1)
test_data = {
    'ID':    [1, 2, 3, 4, 5, 6, 7, 8],
    'Name':  ['Alicia', 'Bob', 'Charlie', 'Dominic', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'Age':   [25, np.nan, 22, 19, 31, 35, np.nan, 18],
    'City':  ['NY', 'LA', 'Chicago', 'Houston', 'Phoenix', 'Boston', 'Austin', 'San Diego'],
    'Score': [85, 92, -1, 65, 78, np.nan, 55, 90]
}
```

```python
# Create DataFrame from test_data
df4 = pd.DataFrame(test_data)
df4
```

```python
# Export df4 to a CSV file (index=False keeps the row numbers out of the file)
df4.to_csv('test_data.csv', index=False)
print('test_data.csv saved successfully.')
```
