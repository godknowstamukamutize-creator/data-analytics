# Exercise 2.C Working with multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
# Creating df_sales from the provided data
data = {
    'Product':    ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region':     ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price':      [100, 200, 150, 100, 200, 150, 100, 200, 150]
}
df_sales = pd.DataFrame(data)
df_sales
```

```python
# Adding revenue column: Units_Sold multiplied by Price
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

```python
# Total Units_Sold and total Revenue across all rows
print('Total Units Sold:', df_sales['Units_Sold'].sum())
print('Total Revenue:   ', df_sales['Revenue'].sum())
```

```python
# Average Units_Sold and average Revenue
print('Average Units Sold:', df_sales['Units_Sold'].mean())
print('Average Revenue:   ', df_sales['Revenue'].mean())
```

## Grouping Data

```python
# Total Units_Sold and Revenue grouped by Product
df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum()
```

```python
# Average Units_Sold and Revenue grouped by Region
df_sales.groupby('Region')[['Units_Sold', 'Revenue']].mean()
```

```python
# Sum of Units_Sold and Revenue grouped by both Product and Region
df_sales.groupby(['Product', 'Region'])[['Units_Sold', 'Revenue']].sum()
```

## Sorting and Ranking Data

```python
# Sorting by Units_Sold ascending
df_sales.sort_values('Units_Sold', ascending=True)
```

```python
# Sorting by Revenue descending
df_sales.sort_values('Revenue', ascending=False)
```

```python
# Adding Revenue_Rank column: rank 1 = highest revenue (ascending=False)
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending=False).astype(int)
df_sales
```

## Merging DataFrames

```python
# Creating df_products with product category information
product_data = {
    'Product':  ['A', 'B', 'C'],
    'Category': ['Electronics', 'Furniture', 'Appliances']
}
df_products = pd.DataFrame(product_data)
df_products
```

```python
# Merging df_sales with df_products on the Product column
df_merged = pd.merge(df_sales, df_products, on='Product')
df_merged
```

```python
# Total Revenue grouped by Category
df_merged.groupby('Category')['Revenue'].sum()
```
